## 📌 Recap

| | `loc` | `iloc` |
|---|---|---|
| Picks by | **label** (name) | **position** (number) |
| Slice end | **included** | **not included** |
| Columns given as | names | numbers |
| Takes a condition | **yes** | no |

**Boolean indexing:** write a condition, get True/False for every row, keep the True ones.
Two conditions need `&` or `|`, with **brackets around each one**.
