## 1️⃣ Introduction to Pandas

> **Flow:** Pandas = NumPy + labels → Series (1D + index) → DataFrame (2D table) → creating from a dict → NumPy sits underneath

```python
! pip3 install pandas
```

```python
# import pandas as pd
import pandas as pd
import numpy as np 
```

```python
# pd.Series() — 1D data with an index
total_bill=[4700,8000,1200,500]
s=pd.Series(total_bill,index=['mon','tue','wed','fri'])
print(s)
print(type(s))
```

```python
# pd.DataFrame() — from a dictionary

restaurant_data = {
    'Table_No': [12, 14, 15],
    'Bill_Amount': [1250, 890, 2400],
    'Waiter_Name': ['Amit', 'Priya', 'Rahul']
}
df = pd.DataFrame(restaurant_data)

# print(df)
# print(type(df))
print(df['Bill_Amount'])
print(type(df['Bill_Amount']))
```

```python
# .values / .to_numpy() — the NumPy array underneath

print(df.values)
print(type(df.values))
```

## 2️⃣ Loading & Inspecting Data

> **Flow:** `pd.read_csv` → `head()` / `tail()` → `shape`, `columns` → `info()` (dtypes + nulls together) → `describe()` → `value_counts()` → `unique()` / `nunique()`

```python

df=pd.read_csv("/Users/shailesh/Desktop/vedam_Sem3_ml/Data/tip.csv")
```

```python
# head(), tail()
# print(df.head(20))
print(df.tail(10))
```

```python
df.sample(5)
```

```python
# shape, columns

df.shape
print(df.columns)
```

```python
# info() — dtypes and null counts in one shot
df.info()
```

```python
# describe() — statistics of numeric columns
df.describe()
```

```python
# value_counts(), unique(), nunique()
print(df['day'].value_counts())
print(df['day'].unique())
print(df['day'].nunique())
```
