## 2️⃣ Creating Arrays

> **Flow:** `np.array()` 1D & 2D → `zeros` / `ones` / `full` → `arange` (stop excluded) vs `linspace` (stop included) → `eye` → `dtype`

````python
# np.array() — 1D and 2D

x=np.array([2,3,4,5])
x1=np.array([[2,3],[4,5],[8,9]])
print(x.dtype)
print(type(x))
print(x1)
````

````python
np.array([1,2.5,"hello"])
x=np.array([1,2.5,5.6])
print(x.dtype)
````

````python
x=np.array([1,4,4.3,4.9,4.8],dtype='int32')
print(x)
````

````python
x=np.array([1,4,4.3,4.9,4.8])
print(x)
y=x.astype('int32')
print(x)
````

## 4️⃣ Indexing & Slicing

> **Flow:** 1D indexing (positive + negative) → slicing `[start:stop:step]` → reverse trick `[::-1]` → 2D `[row, col]` (row / column / block) → boolean indexing `arr[arr > 50]` (the foundation of Pandas filtering)

````python
# 1D indexing — positive and negative
arr=np.arange(10)
print(arr)
print(arr[-3:-12:-2])
````

````python
# 1D slicing — [start:stop:step], reverse
````

````python
# 2D indexing — arr[row, col], full row, full column

arr=np.arange(20).reshape(5,4)
print(arr)
print(arr[2,2])
````

````python
# 2D slicing — sub-matrix (block)
print(arr)
# print(arr[2:,1:])
# print(arr[1,1:])

# print(arr[::4,::3])
print(arr[1:3,2::-1])
````

````python
# Boolean indexing — arr[condition]
````

## 5️⃣ Basic Operations (Element-wise)

> **Flow:** array + scalar → array + array → `+ - * / **` → comparison operators → boolean array

````python
# arithmetic — array with scalar, array with array
# arr1=np.arange(10)
# print(arr1)
# arr1=arr1+5
# print(arr1)
# print("Multiply",arr1*5)
# print("exponent",arr1**2)


x=np.arange(10)
y=np.arange(11,21)
print(x)
print(y)
print(x*y)
````

````python
x=np.arange(6).reshape(2,3)
y=np.arange(3)
print("x is",x)
print("y is",y)
print(x+y)
````

````python
x=np.arange(6).reshape(2,3)
y=np.arange(6).reshape(3,2)
print("x is",x)
print("y is",y)
print(x+y)
````

````python
# comparison operators — boolean arrays
x=np.arange(5)
y=np.arange(1,6)
print(x)
print(y)
print(x==y)
print(x>y)
print(y>x)
````

## 6️⃣ The `np.random` Module

> **Flow:** `seed` (why reproducibility matters) → `rand` → `randn` → `randint` (high excluded) → `choice` → `shuffle`

````python
# np.random.seed() + np.random.rand()
np.random.seed(2)
np.random.rand(5)
np.random.randn(10)
````

````python
# np.random.randint(low, high, size)
np.random.randint(10,20,size=(5,3))
````

````python
import numpy as np 
````

````python
# np.random.choice(), np.random.shuffle()
lst=[3,4,56,7,9]
np.random.choice(lst,3,replace=False)
np.random.choice(10,5,replace=False)
````

````python
arr=np.arange(10)
np.random.shuffle(arr)
print(arr)
````
